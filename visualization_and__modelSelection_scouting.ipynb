```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("Scouting_Database_Final.csv")
df.head()
```

So this first plot will show the relationship between shots/90 and goals/90 with interactive hover

```python
fig = px.scatter(
    df,
    x="standard_sh_per90",
    y="performance_gls_per90",
    hover_name="player",
    hover_data=["team", "league", "pos", "age"],
    color = "pos",
    title = "Goals per 90 vs. Shots per 90",
    labels = {
        "standard_sh_per90": "Shots per 90",
        "performance_gls_per90": "Goals per 90"
    }
)

fig.show()
```

So as you can see from that scatter plot, there is a positive correlation between shots and goals scored

The second visualization I'm going to do is a correlation map

```python
#first i select the numeric columns
numeric_cols = [col for col in df.columns if col.endswith("_per90") or col.endswith("%")]
corr = df[numeric_cols].corr()

fig2 = px.imshow(
    corr,
    title = "Feature Correlation matrix",
    color_continuous_scale="RdBu",
    labels=dict(x="Features", y="Features", color="Correlation")
)

fig2.show()
```

**`MODEL SELECTION`**

So i have decided to go with a model that will use the Cosine similarity algorithm.
Why:
1. Our goal is not predicting a single categorical label or numerical value. Instead, I aim to rank player profiles based on how closely their playstyle mirrors a target player.
2. Cosine Similarity calculates the angle between two feature vectors. This measures the shape of a player's output (their specific proportion of passes, tackles, and shots) rather than simply rewarding raw volume.

`FEATURE SCALING`

I will use StandardScaler because Cosine Similarity measures the angle between two feature vectors, and if features are on drastically different numerical scales, the larger feature will overpower the smaller ones and skew the angle.
It will preserve relative variance and it handles outliers.


If we look at our dataset we have features like standard_sot%  (values range from 0 to 100) and goals per 90 (0.0 to 0.8)
Without scaling, the standard shots on target feature will dominate the goals per 90 feature.
